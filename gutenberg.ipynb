```python
import pandas as pd

authors = pd.read_csv("data/gutenberg_authors.csv")
metadata = pd.read_csv("data/gutenberg_metadata.csv")
languages = pd.read_csv("data/gutenberg_languages.csv")
```

```python
author_metadata = authors.merge(
    metadata,
    on="gutenberg_author_id"
)

author_metadata.head()
```

```python
full_data = author_metadata.merge(
    languages,
    on="gutenberg_id"
)

full_data.head()
```

```python
full_data.head()
```

```python
full_data.columns
```

```python
full_data[
    ["alias", "title", "language_y", "total_languages"]
].head(10)
```

```python
full_data[
    full_data["total_languages"] > 1
][["alias", "title", "language_y", "total_languages"]].head(20)
```

```python
full_data[
    ["alias", "title", "total_languages"]
].sort_values(
    "total_languages",
    ascending=False
).head(20)
```

```python
clean_data = full_data.dropna(subset=["alias"])
```

```python
clean_data[
    ["alias", "title", "total_languages"]
].sort_values(
    "total_languages",
    ascending=False
).head(20)
```

```python
author_counts = (
    clean_data
    .groupby("alias")
    .size()
    .sort_values(ascending=False)
)

author_counts.head(20)
```

```python
clean_data[
    clean_data["alias"] == "Wells, Herbert George"
][["title", "language_y", "total_languages"]].head(20)
```

```python
clean_data[
    (clean_data["alias"] == "Wells, Herbert George")
    & (clean_data["total_languages"] > 1)
][["title", "language_y", "total_languages"]]
```

```python
translated = clean_data[
    clean_data["total_languages"] > 1
]

translated[
    ["alias", "title", "language_y", "total_languages"]
].head(20)
```

```python
translated_counts = (
    translated
    .groupby("alias")
    .size()
    .sort_values(ascending=False)
)

translated_counts.head(20)
```

```python
translated_counts.head(20).to_frame("count")
```

```python
translated[
    translated["alias"] == "Erasmus, of Rotterdam"
][["title", "language_y", "total_languages"]]
```

```python
translated_counts = (
    translated
    .groupby("alias")
    .size()
    .sort_values(ascending=False)
)

translated_counts
```

```python
works = clean_data[
    ["alias", "gutenberg_id", "total_languages"]
].drop_duplicates()
```

```python
works["translations"] = works["total_languages"] - 1
```

```python
translation_counts = (
    works
    .groupby("alias")["translations"]
    .sum()
    .sort_values(ascending=False)
)
```
